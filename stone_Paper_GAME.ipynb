```python
import random

dic = {"p": 0, "c": -1, "s": 1}
rev = {0: "paper", -1: "pencil", 1: "stone"}
sc=0;
for i in range(3):
    comp = random.choice([0, -1, 1])
    youS = input("Enter s (stone), p (paper), c (pencil): ").lower()

    if youS not in dic:
        print("Invalid choice!")
        continue

    you = dic[youS]
    print(f"You chose {rev[you]}\nComputer chose {rev[comp]}")

    if you == comp:
        print("DRAW")
    elif you - comp in [-1, 2]:
        print("🏆 You win!")
        sc=sc+1
    else:
        print("You lose!")

    print(f"Round {i + 1} completed\n")

print("Your score is:",sc)
```
